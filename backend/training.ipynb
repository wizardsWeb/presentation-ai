### Subtask:
Demonstrate how to load the fine-tuned model and perform inference on new input prompts to evaluate its performance in generating NCERT-based explanations. Provide example prompts and compare outputs to assess the quality of the fine-tuning.


**Reasoning**:
I need to load the base model and tokenizer, then load the fine-tuned LoRA adapters from the specified directory, merge them into the base model, and then perform inference on new input prompts as instructed. This requires importing specific modules from `transformers` and `peft`.



````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from peft import PeftModel

# Clear GPU memory if any previous models are still loaded
torch.cuda.empty_cache()
print("GPU memory cleared for fresh inference model loading.")

# Define the base model ID
base_model_id = "microsoft/phi-2"

# Quantization configuration (must match training)
bitsandbytes_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=False,
)

# Load base model
base_model = AutoModelForCausalLM.from_pretrained(
    base_model_id,
    quantization_config=bitsandbytes_config,
    trust_remote_code=True,
    device_map="auto"
)

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(base_model_id, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

print(f"Base model '{base_model_id}' and tokenizer loaded.")

# Load the saved LoRA adapters
fine_tuned_model_path = './fine_tuned_phi_model'
model = PeftModel.from_pretrained(base_model, fine_tuned_model_path)
print(f"LoRA adapters loaded from '{fine_tuned_model_path}'.")

# Merge LoRA adapters into the base model
model = model.merge_and_unload()
print("LoRA adapters merged into the base model.")

# Define new input prompts for inference (not in training set)
new_input_prompts = [
    "What is the function of the human brain according to NCERT biology?",
    "Explain the causes and effects of the Green Revolution in India as per NCERT Social Science.",
    "Describe the characteristics of sound waves, based on NCERT Physics.",
    "What is the importance of forests according to NCERT Geography?"
]

print("Starting inference with new prompts...")

for i, prompt in enumerate(new_input_prompts):
    print(f"\n--- Inference for Prompt {i+1} ---")
    print(f"Instruction: {prompt}")

    # Construct the input string for the model
    input_string = f"### Instruction:\n{prompt}\n### Response:\n"

    # Encode the input string
    inputs = tokenizer(input_string, return_tensors="pt").to(base_model.device)

    # Generate text
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=500, # Increased max_new_tokens for potentially longer responses
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            eos_token_id=tokenizer.eos_token_id
        )

    # Decode the generated tokens
    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Extract only the response part, removing the instruction and the initial part of the response prompt
    response_start_tag = "### Response:\n"
    response_start_index = generated_text.find(response_start_tag)

    if response_start_index != -1:
        extracted_response = generated_text[response_start_index + len(response_start_tag):].strip()
        print(f"Generated Response:\n{extracted_response}")
    else:
        print(f"Generated Output (could not parse response):\n{generated_text}")

print("\nInference complete. Review the generated responses above.")
````

````python
import torch
import gc
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, StoppingCriteria, StoppingCriteriaList

# Define a custom stopping criteria to stop generation at various instruction patterns
class StopOnMultipleInstructions(StoppingCriteria):
    def __init__(self, tokenizer, stop_sequences_str_list):
        self.tokenizer = tokenizer
        self.stop_sequences_ids = []
        for seq_str in stop_sequences_str_list:
            encoded_seq = tokenizer.encode(seq_str, add_special_tokens=False, return_tensors='pt').squeeze()
            if encoded_seq.dim() == 0:
                encoded_seq = encoded_seq.unsqueeze(0)
            self.stop_sequences_ids.append(encoded_seq)

    def __call__(self, input_ids: torch.LongTensor, scores: torch.FloatTensor, **kwargs) -> bool:
        for stop_seq_ids in self.stop_sequences_ids:
            stop_seq_ids_on_device = stop_seq_ids.to(input_ids.device)
            seq_len = stop_seq_ids_on_device.shape[0]

            if input_ids.shape[1] >= seq_len:
                if torch.equal(input_ids[0, -seq_len:], stop_seq_ids_on_device):
                    return True
        return False

# Clear GPU memory and collect garbage (important for local reloading)
torch.cuda.empty_cache()
gc.collect()
print("GPU memory cleared and garbage collected.")

# Define the path to your locally saved merged model
local_model_path = './merged_fine_tuned_phi_model'

# Quantization configuration (must match training)
bitsandbytes_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=False,
)

# Load the merged model and tokenizer
print(f"Loading merged model from {local_model_path}...")
model = AutoModelForCausalLM.from_pretrained(
    local_model_path,
    quantization_config=bitsandbytes_config,
    trust_remote_code=True,
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(local_model_path, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

print("Merged model and tokenizer loaded successfully.")

# Define multiple stop sequences, including variations for exercises, problems, and standalone '#' symbol
stop_sequences_str_list = [
    "### Instruction:", "## Instruction:", "# Instruction:",
    "# Exercise", "## Exercise:",
    "### Response:",
    "# Problem", "# Solution",
    "#",
    "Question:",
    "Possible rewrite of the question and solution:"
]

# Initialize stopping criteria
stop_criteria = StoppingCriteriaList([StopOnMultipleInstructions(tokenizer, stop_sequences_str_list)])

def generate_ncert_explanation(topic: str) -> str:
    """
    Generates an NCERT-based explanation for a given topic using the loaded fine-tuned model.

    Args:
        topic (str): The topic or question to explain.

    Returns:
        str: The generated NCERT-based explanation.
    """
    input_string = f"### Instruction:\n{topic}\n### Response:\n"
    inputs = tokenizer(input_string, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=500,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            eos_token_id=tokenizer.eos_token_id,
            stopping_criteria=stop_criteria
        )

    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    response_start_tag = "### Response:\n"
    response_start_index = generated_text.find(response_start_tag)

    if response_start_index != -1:
        extracted_response = generated_text[response_start_index + len(response_start_tag):].strip()

        # Further refine to remove any partial stop sequences that might have been generated just before stopping
        min_idx = len(extracted_response)
        for seq in stop_sequences_str_list:
            idx = extracted_response.find(seq)
            if idx != -1:
                min_idx = min(min_idx, idx)
        extracted_response = extracted_response[:min_idx].strip()

        return extracted_response
    else:
        return "Could not parse generated response."

print("\n'generate_ncert_explanation' function defined.")

# Test the function with new prompts
print("\n--- Testing inference function with new prompts ---")

example_prompts = [
    "What are the three states of matter according to NCERT Science?",
    "Explain the causes of the French Revolution as per NCERT History.",
    "Describe the main types of ecosystems mentioned in NCERT Biology."
]

for i, eprompt in enumerate(example_prompts):
    print(f"\nExample Prompt {i+1}: {eprompt}")
    response = generate_ncert_explanation(eprompt)
    print(f"Generated Explanation:\n{response}")

print("\nInference testing complete.")
````
